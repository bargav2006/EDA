```python
import numpy as np
import pandas as pd

ind=pd.Index([2,3,5,7,11])
print(ind)

print(ind.size,ind.shape,ind.ndim,ind.dtype)


print(ind[1])
print(ind[::2])
print(ind[1:4:2])

indA=pd.Index([1,3,5,7,9])
indB=pd.Index([2,3,5,7,11])

print(indA & indB)
print(indA | indB)
print(indA ^ indB)
print(indA.intersection(indB))

```

```python
import pandas as pd

index = [('A', 2010), ('A', 2020), ('A', 2025), ('B', 2010), ('B', 2020), ('B', 2025)]
population = [123, 456, 789, 101, 121, 131]

pop = pd.Series(population)

i = pd.MultiIndex.from_tuples(index)
print(i)

pop = pop.reindex(i)
print(pop)

pop_df = pop.unstack()
print(pop_df)
pop_df.stack()
```

```python
import pandas as pd


states = ['A', 'A', 'A', 'B', 'B', 'B']
years = [2010, 2020, 2025, 2010, 2020, 2025]
population = [123, 456, 789, 101, 121, 131]
i = pd.MultiIndex.from_arrays([states, years], names=['state', 'year'])
print(i)

pop = pd.Series(population, index=i)
print(pop)
pop_df = pop.unstack()
print(pop_df)
```

```python
import pandas as pd

states = ['A', 'B']
years = [2010, 2020, 2025]
population = [123, 456, 789, 101, 121, 131]

i = pd.MultiIndex.from_product([states, years], names=['state', 'year'])
print(i)

pop = pd.Series(population, index=i)
print(pop)
pop_df = pop.unstack()
print(pop_df)
```

```python
import pandas as pd

products = {
    'Laptop': 999.99,
    'Mouse': 25.50,
    'Keyboard': 45.00
}

price_series = pd.Series(products)

print("Pandas Series:")
print(price_series)

index_as_list = list(price_series.index)
dict_keys = list(products.keys())

is_matching = index_as_list == dict_keys

print("\nSeries Index:", index_as_list)
print("Dictionary Keys:", dict_keys)
print("Do they match?", is_matching)
```

```python
import pandas as pd

area = pd.Series(
    [163696, 268596, 141297, 54556, 65758],
    index=['California', 'Texas', 'Montana', 'New York', 'Florida']
)

population = pd.Series(
    [38980000, 30500000, 1130000, 19570000, 22610000],
    index=['California', 'Texas', 'Montana', 'New York', 'Florida']
)

states = pd.DataFrame({'area': area, 'population': population})

states['density'] = states['population'] / states['area']

print(states)
```
