```python
import numpy as np

my_list = [10, 20, 30, 40, 50]

my_array = np.array(my_list)

print("Array:", my_array)


print("Type:", type(my_array))
```

```python
import numpy as np

my_matrix = np.arange(1, 7).reshape(2, 3)

print("Matrix:\n", my_matrix)

print("Type:", type(my_matrix))
```

```python
import numpy as np
arr = np.empty((2, 2))

print("Array:\n", arr)
print("Type:", type(arr))

# np.empty creates an array without initializing its values.
# That means the memory is allocated, but the contents are whatever
# happens to already be in that memory location so the unpredictable values.
#
# 
# np.empty is faster because it skips the step of filling the array with zeros.
# It's useful when you plan to overwrite all elements immediately,
# 
```
